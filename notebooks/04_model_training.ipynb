### 📊 Öne Çıkan Özellikler (Feature Importance)

Model eğitimi sonucunda, üretilen özellikler arasında aşağıdaki değişkenler ilk 30 içerisinde yer almıştır:

| Özellik | Açıklama |
|----------|----------|
| credit_term | Kredi tutarı / taksit tutarı |
| annuity_credit_ratio | Taksit tutarı / kredi tutarı |
| bureau_total_credit | Geçmiş toplam kredi hacmi |
| bureau_total_debt | Geçmiş toplam borç |
| bureau_debt_credit_ratio | Borç / kredi oranı |
| bureau_active_loans | Aktif kredi sayısı |
| prev_avg_credit_amount | Önceki başvurulardaki ortalama kredi |
| prev_avg_application_amount | Önceki başvurulardaki ortalama talep |
| approval_rate | Önceki başvuruların onay oranı |
| refusal_rate | Önceki başvuruların reddedilme oranı |
| total_payment_amount | Toplam geçmiş ödeme tutarı |
| late_payment_ratio | Gecikmeli ödeme oranı |
| avg_days_late | Ortalama gecikme günü |
| installment_count | Toplam taksit sayısı |
| avg_payment_ratio | Ortalama ödeme/taksit oranı |
